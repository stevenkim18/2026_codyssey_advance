## 0.3 전체 파이프라인

문장 하나가 숫자 벡터가 되어 검색과 분류에 쓰이기까지의 과정입니다.

```
 원문            "NASA launched a new Space Telescope!"
   │ ① 정제 (8장): 소문자로 바꾸고 특수문자를 지운다
   ▼
                 "nasa launched a new space telescope"
   │ ② 토큰화 (9장): 단어 단위로 자른다
   ▼
                 ["nasa", "launched", "a", "new", "space", "telescope"]
   │ ③ 불용어 제거 (10장): 의미가 약한 단어를 뺀다
   ▼
                 ["nasa", "launched", "space", "telescope"]
   │ ④ 벡터화 (11–14장): 어휘 크기만큼의 숫자 배열로 바꾼다 (BoW → TF-IDF)
   ▼
                 [0.00, 0.41, 0.00, 0.52, ...]
   ├──▶ ⑤ 검색 (15–16장): 쿼리 벡터와 코사인 유사도를 계산해 상위 5개 반환
   └──▶ ⑥ 분류 (17–20장): 벡터를 입력으로 주제를 예측하고 정확도·F1로 평가
```

### 미리 보기: 30줄로 만든 초소형 검색기
아래 코드는 이 과제 전체를 아주 작게 줄인 것입니다. **지금 코드를 이해하지 못해도 괜찮습니다.** 주석의 장 번호에서 하나씩 배웁니다. 실행해서 "결국 이런 걸 만드는구나"만 확인하세요.

````python
import re
from collections import Counter

import numpy as np

docs = [
    "The rocket launched into space!",
    "NASA launched a new space telescope.",
    "The pitcher threw a fast ball.",
    "Our team won the baseball game.",
    "Astronauts watched the baseball game in space.",
]
stopwords = {"the", "a", "into", "our", "in", "new"}


def preprocess(text):
    text = text.lower()                                # ① 정제: 소문자 (8장)
    text = re.sub(r"[^a-z\s]", " ", text)              # ① 정제: 영문자와 공백만 남김 (8장)
    tokens = text.split()                              # ② 토큰화: 공백 기준 (9장)
    return [t for t in tokens if t not in stopwords]   # ③ 불용어 제거 (10장)


tokenized = [preprocess(d) for d in docs]
vocab = sorted({t for doc in tokenized for t in doc})  # 어휘 사전 (11장)
index = {w: i for i, w in enumerate(vocab)}


def count_vector(tokens):                              # ④ BoW 카운트 (11장)
    v = np.zeros(len(vocab))
    for t, c in Counter(tokens).items():
        if t in index:
            v[index[t]] = c
    return v


counts = np.array([count_vector(t) for t in tokenized])
df = (counts > 0).sum(axis=0)                          # ④ 문서 빈도 (12장)
n = len(docs)
idf = np.log((1 + n) / (1 + df)) + 1                   # ④ smooth IDF (12장)


def tfidf(c):                                          # ④ TF-IDF와 L2 정규화 (12–13장)
    v = c * idf
    norm = np.linalg.norm(v, axis=-1, keepdims=True)
    return v / np.where(norm == 0, 1, norm)


X = tfidf(counts)
print("어휘 수:", len(vocab), "| 문서-단어 행렬 shape:", X.shape)

query = "space rocket"
q = tfidf(count_vector(preprocess(query)))
scores = X @ q                                         # ⑤ 코사인 유사도 (15장)
print(f'Query: "{query}"')
for rank, i in enumerate(np.argsort(scores)[::-1][:3], start=1):  # ⑤ 상위 k개 (16장)
    print(f"  {rank}. [{scores[i]:.3f}] doc {i}: {docs[i]}")
````

**결과 읽기**
- 문서 5개가 `(5, 어휘 수)` 모양의 숫자 표(행렬)가 되었습니다. 한 행이 문서 하나입니다.
- `space`와 `rocket`을 모두 가진 doc 0이 1등입니다.
- doc 1과 doc 4는 `space`만 겹쳐서 점수가 낮습니다. doc 4는 우주비행사가 *야구 경기를 보는* 이야기인데, 로켓과는 상관없이 `space`라는 단어 하나 때문에 검색되었습니다. **단어만 보고 문맥은 보지 않는 BoW 방식**의 특징이 벌써 보입니다(21장).
